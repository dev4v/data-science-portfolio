```python

#take 2011-2024

import pandas as pd



# Load the dataset
df = pd.read_csv('Electric_Vehicle_Population_Data.csv')

# Group the data by 'Model Year' and calculate the number of EVs sold per year
year_sales = df.groupby('Model Year').size().reset_index(name='Sales Count')

# Filter the data for Model Years between 2011 and 2024
year_sales_filtered = year_sales[(year_sales['Model Year'] >= 2011) & (year_sales['Model Year'] <= 2024)]

# Sort the filtered data by Model Year
year_sales_filtered = year_sales_filtered.sort_values(by='Model Year', ascending=True)
```

```python
# Plot the correlation map
import seaborn as sns

correlation_matrix=year_sales_filtered.corr()
plt.figure(figsize=(10, 8))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm')
plt.title("Correlation Map")
plt.show()
```

```python
from scipy import stats

mean_value = year_sales_filtered['Sales Count'].mean()
print("mean",mean_value)

median_value=year_sales_filtered['Sales Count'].median()
print("median",median_value)


mode_value=year_sales_filtered['Sales Count'].mode(0)[0]
print("mode",mode_value)

variance_value=year_sales_filtered['Sales Count'].var()
print("variance",variance_value)

std_dev_value=year_sales_filtered['Sales Count'].std()
print("std_dev_value",std_dev_value)


q1 = year_sales_filtered['Sales Count'].quantile(0.25)  # 25th percentile
q3 = year_sales_filtered['Sales Count'].quantile(0.75)  # 75th percentile
iqr = q3 - q1  
print("IQR (Interquartile Range)",iqr)

range=max(year_sales_filtered['Sales Count'])-min(year_sales_filtered['Sales Count'])
print("range",range)


coeff_variance=std_dev_value/mean_value
print("coefficient of variance :",coeff_variance)

coeff_std_dev=std_dev_value/mean_value
print("coefficient of standard deviation:",coeff_std_dev)
```

```python
import pandas as pd

# Loading dataset
df = pd.read_csv('Electric_Vehicle_Population_Data.csv')

# Group the data by 'Model Year' and calculate the number of EVs sold per year
year_sales = df.groupby('Model Year').size().reset_index(name='Sales Count')

# Filter the data for Model Years between 2011 and 2024
year_sales_filtered = year_sales[(year_sales['Model Year'] >= 2011) & (year_sales['Model Year'] <= 2024)]

# Sort the filtered data by Model Year
year_sales_filtered = year_sales_filtered.sort_values(by='Model Year', ascending=True)

# Calculate the IQR (Interquartile Range)
q1 = year_sales_filtered['Sales Count'].quantile(0.25)  # 25th percentile
q3 = year_sales_filtered['Sales Count'].quantile(0.75)  # 75th percentile
iqr = q3 - q1  # Interquartile range

# Print the filtered data and IQR
print(year_sales_filtered)
#print(f"IQR (Interquartile Range): {iqr}")
```

```python
plt.figure(figsize=(10, 6))  # Set the figure size
sb.lineplot(x='Model Year', y='Sales Count', data=year_sales_filtered, marker='o')
plt.title('EV Sales from 2011 to 2024')  # Title of the graph
plt.xlabel('Year')  # Label for the X-axis
plt.ylabel('Sales Count')  # Label for the Y-axis
plt.xticks(year_sales_filtered['Model Year'])  # Ensures every year is shown on the x-axis
plt.grid(True)  # Add a grid for better readability
plt.show()  # Display the plot
```

```python
# Calculate the mean of 'Sales Count' per 'Model Year'
mean_sales_per_year = year_sales_filtered.groupby('Model Year')['Sales Count'].mean().reset_index()

# Check the mean values
print(mean_sales_per_year)
```

```python
import matplotlib.pyplot as plt

# Plot a histogram of the mean sales counts
plt.figure(figsize=(10, 6))

#plt.hist(mean_sales_per_year['Sales Count'], bins=10, edgecolor='k', color='skyblue')
sb.histplot(mean_sales_per_year['Sales Count'],bins=10,color='blue',edgecolor='k',stat='density')
sb.kdeplot(mean_sales_per_year['Sales Count'],color='red')
plt.title('Histogram of Mean EV Sales Count')
plt.xlabel('Mean Sales Count')
#plt.ylabel('Frequency')
plt.grid(True)
plt.show()
```

```python
#KDE plot 

plt.figure(figsize=(8,10))
sb.kdeplot(mean_sales_per_year['Sales Count'],fill=True,color='red')
plt.title("density plot")
plt.xlabel("mean sales count")
plt.show()
```

```python
import plotly.express as px
import pandas as pd
import json
import requests

# Load GeoJSON for Washington counties
url = "https://raw.githubusercontent.com/plotly/datasets/master/geojson-counties-fips.json"
counties = requests.get(url).json()

# Filter for only Washington counties
wa_counties = [county for county in counties['features'] if county['properties']['STATE'] == '53']
wa_geojson = {'type': 'FeatureCollection', 'features': wa_counties}


county_name_map = {
    'Grays Harbor': 'Grays Harbor County',
    # Add more mappings if needed
}

county_ev_data_new['County'] = county_ev_data_new['County'].replace(county_name_map)

# Create the choropleth map
fig = px.choropleth(county_ev_data_new, 
                    geojson=wa_geojson, 
                    locations='County', 
                    color='EV_Population',
                    color_continuous_scale="Viridis",
                    featureidkey="properties.NAME",
                    scope="usa",
                    labels={'EV_Population':'EV Population'},
                    title='EV Population by County in Washington')

# Update the layout to zoom in on Washington
fig.update_geos(fitbounds="locations", visible=False)

# Adjust the size of the figure
fig.update_layout(height=600, width=800)

# Show the map
fig.show()
```
